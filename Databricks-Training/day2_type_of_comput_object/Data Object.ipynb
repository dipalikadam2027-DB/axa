```sql
CREATE OR REPLACE FUNCTION dev.dipali.mask_email(email STRING)
RETURNS STRING
RETURN CONCAT(
  LEFT(email, 1),
  REPEAT('*', INSTR(email, '@') - 2),
  SUBSTRING(email, INSTR(email, '@') - 1)
);
```

```sql
select *, dev.dipali.mask_email(email_address) as mask_emails from samples.bakehouse.sales_customers
```
