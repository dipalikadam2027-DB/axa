increamental : Data which grows over time (per hour, per day and so on) -- COPY into, Pipeline
Streaming : Data which grows over time (per sec, per min and so on) -- Structured, Autoloader , Pipeline

idempotnent(skip the file which read previously and it should read only new data ) 

Structured Streaming(python, spark) -- COPY INTO (SQL)

```python
%python
# Batch
df=spark.read.csv("/Volumes/dev/naval/raw/months/",header=True,inferSchema=True)
df.write.mode("append").saveAsTable("dev.dipali_bronze.sales_new")
```

```python
%sql

df=spark.readStream.csv("/Volumes/dev/naval/raw/months/",header=True,inferSchema=True)
df.writeStream.mode("append").saveAsTable("dev.dipali_bronze.sales_new")
```
