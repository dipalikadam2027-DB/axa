```sql
%python
dbutils.widgets.text("catalog","")
dbutils.widgets.text("table","")
catalog=dbutils.widgets.get("catalog")
table=dbutils.widgets.get("table")
```

```sql
%python
from pyspark.sql.functions import *
input_path="/Volumes/dev/naval_bronze/raw/"
def add_ingestion_metadata(df):
  return df.withColumn("ingestion_date",current_timestamp()).withColumn("file_name",col("_metadata.file_name")).withColumn("last_modified_ts",col("_metadata.file_modification_time")).withColumn("file_path",col("_metadata.file_path"))
df=spark.read.csv(f"{input_path}{table}.csv",header=True,inferSchema=True)

df1=add_ingestion_metadata(df)

df1.write.mode("overwrite").saveAsTable(f"{catalog}.dipali_bronze.{table}")
```
